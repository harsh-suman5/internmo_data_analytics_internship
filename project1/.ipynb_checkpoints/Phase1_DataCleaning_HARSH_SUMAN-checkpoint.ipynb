step 4

```python
datasets = {
    "Subscribers": subscribers,
    "Titles": titles,
    "Watch History": watch_history,
    "Ratings": rating,
    "Reviews": reviews,
    "Watchlist": watchlist
}

print("=" * 70)
print("DUPLICATE ROWS REPORT")
print("=" * 70)

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
primary_keys = {
    "Subscribers": ("subscriber_id", subscribers),
    "Titles": ("title_id", titles),
    "Watch History": ("watch_id", watch_history),
    "Ratings": ("rating_id", rating),
    "Reviews": ("review_id", reviews),
    "Watchlist": ("watchlist_id", watchlist)
}

print("=" * 70)
print("PRIMARY KEY DUPLICATE REPORT")
print("=" * 70)

for name, (key, df) in primary_keys.items():

    if key in df.columns:
        duplicate_key = df[key].duplicated().sum()
        print(f"{name} ({key}): {duplicate_key} duplicate IDs")
    else:
        print(f"{name}: '{key}' column not found")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(
    watch_history["watch_duration_min"].dropna(),
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Watch Duration", fontsize=16)
plt.xlabel("Watch Duration (minutes)", fontsize=12)
plt.ylabel("Number of Sessions", fontsize=12)

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# iqr method

sorted(watch_history['watch_duration_min'])

q1,q3 = np.percentile(watch_history['watch_duration_min'], [25,75])
print(q1,q3)
```

```python
iqr = q3-q1
print(iqr)
```

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
print(lower_bound, upper_bound)
```

```python
outliers = watch_history[
    (watch_history["watch_duration_min"] < lower_bound) |
    (watch_history["watch_duration_min"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

```python
outliers[
    [
        "watch_id",
        "title_id",
        "content_duration_min",
        "watch_duration_min"
    ]
].head(20)
```
