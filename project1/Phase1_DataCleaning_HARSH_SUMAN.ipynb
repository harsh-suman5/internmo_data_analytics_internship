```python
# iqr method

sorted(watch_history['watch_duration_min'])

q1,q3 = np.percentile(watch_history['watch_duration_min'], [25,75])
print(q1,q3)
```

```python
iqr = q3-q1
print(iqr)
```

```python
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
print(lower_bound, upper_bound)
```

```python
outliers = watch_history[
    (watch_history["watch_duration_min"] < lower_bound) |
    (watch_history["watch_duration_min"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

# Step 8 — Data Quality Report

## Summary of Findings

### 1. Dataset Overview

Six interconnected datasets were loaded and inspected using Python and Pandas: 
`subscribers.csv`, `titles.csv`, `watch_history.csv`, `ratings.csv`, 
`reviews.csv`, and `watchlist.csv`.
### 2. Basic datasets profilling
The datasets were profiled for their structure, data types, missing values, 
duplicate records, date fields, and data-quality issues. The `watch_history.csv`
table contains 650,000 viewing-session records and acts as the central table 
for viewing activity.

### 3. Missing Value Findings

Missing-value analysis was performed for every column across all six datasets.

The main missing-value findings included:

- `subscribers.csv` contains missing values in `churn_date`. This is expected 
  for subscribers who have not churned and therefore requires interpretation 
  rather than automatic filling.
- `titles.csv` contains missing values in `license_expiry`.
- Other columns with missing values were identified during the missing-value 
  profiling step and were documented with their respective percentages.

Missing values were not automatically filled or deleted during this phase. 
They were identified and documented for appropriate treatment in subsequent 
analysis/cleaning steps.

### 4. Duplicate Record Findings

Duplicate checks were performed on the primary identifiers of all six datasets.

The following duplicate counts were obtained:

- `subscribers.csv` — `subscriber_id`: 0 duplicate IDs
- `titles.csv` — `title_id`: 0 duplicate IDs
- `watch_history.csv` — `watch_id`: 0 duplicate IDs
- `ratings.csv` — `rating_id`: 0 duplicate IDs
- `reviews.csv` — `review_id`: 0 duplicate IDs
- `watchlist.csv` — `watchlist_id`: 0 duplicate IDs

Therefore, no duplicate primary-key IDs were detected in the datasets.

### 5. Data Type and Date Validation Findings

The datasets were inspected for appropriate data types. Date-related columns 
such as `watch_date`, `signup_date`, `churn_date`, `rating_date`, and other 
date fields were reviewed and converted/validated where required.

Numeric fields such as `watch_duration_min`, `content_duration_min`, 
, and monetary fields were also checked for appropriate 
numeric representation.

Date parsing required attention because some date columns were initially 
represented as object/string types. After conversion, date values were 
handled as datetime values for further validation.

### 6. Outlier Findings

Outlier analysis was performed on the `watch_duration_min` column using 
statistical outlier detection and visualization.

The analysis identified several unusually large watch-duration values. 
Examples include values such as 1,866.4 minutes, 1,148.0 minutes, 
2,630.4 minutes, 1,682.5 minutes, and other high-duration observations.

A boxplot and histogram were also used to visually inspect the distribution 
of `watch_duration_min`.

However, the business-rule validation showed that there were **0 sessions**
where:

`watch_duration_min > content_duration_min`

Therefore, although statistical outliers were detected, no viewing session 
was found to have a watch duration greater than its corresponding content 
duration.

The identified statistical outliers were therefore flagged for investigation 
rather than automatically removed.

### 7. Referential Integrity Findings

Referential-integrity checks were performed/are being performed to verify 
that IDs in child tables correctly link to their parent tables.

The following relationships were checked:

- `watch_history.subscriber_id` → `subscribers.subscriber_id`
- `watch_history.title_id` → `titles.title_id`
- `ratings.subscriber_id` → `subscribers.subscriber_id`
- `ratings.title_id` → `titles.title_id`
- `reviews.subscriber_id` → `subscribers.subscriber_id`
- `reviews.title_id` → `titles.title_id`
- `watchlist.subscriber_id` → `subscribers.subscriber_id`
- `watchlist.title_id` → `titles.title_id`

No orphan subscriber IDs or title IDs were identified during the referential
integrity checks. Therefore, the relationships between the relevant tables
were found to be consistent.

### 8. Overall Data Quality Conclusion

The Phase 1 data-quality assessment successfully profiled the six datasets
and identified missing values, validated duplicate primary-key records,
reviewed data types and date fields, investigated statistical outliers in
viewing duration, and checked referential integrity between related tables.

No duplicate primary-key IDs were detected. Statistical outliers were found
in `watch_duration_min`, but no session exceeded its corresponding
`content_duration_min`. The referential-integrity checks also found no orphan
subscriber IDs or title IDs in the related datasets.

Missing values were documented rather than automatically modified because
their appropriate treatment depends on the meaning of each field. The
identified data-quality findings will be considered during the subsequent
cleaning, EDA, and KPI analysis phases.
