step 4

```python
datasets = {
    "Subscribers": subscribers,
    "Titles": titles,
    "Watch History": watch_history,
    "Ratings": rating,
    "Reviews": reviews,
    "Watchlist": watchlist
}

print("=" * 70)
print("DUPLICATE ROWS REPORT")
print("=" * 70)

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()
    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
primary_keys = {
    "Subscribers": ("subscriber_id", subscribers),
    "Titles": ("title_id", titles),
    "Watch History": ("watch_id", watch_history),
    "Ratings": ("rating_id", rating),
    "Reviews": ("review_id", reviews),
    "Watchlist": ("watchlist_id", watchlist)
}

print("=" * 70)
print("PRIMARY KEY DUPLICATE REPORT")
print("=" * 70)

for name, (key, df) in primary_keys.items():

    if key in df.columns:
        duplicate_key = df[key].duplicated().sum()
        print(f"{name} ({key}): {duplicate_key} duplicate IDs")
    else:
        print(f"{name}: '{key}' column not found")
```

```python
datasets = {
    "subscribers": subscribers,
    "titles": titles,
    "watch_history": watch_history,
    "ratings": rating,
    "reviews": reviews,
    "watchlist": watchlist
    }

for name, df in datasets.items():
    print(f"\n{name} dataset:")
    print(df.describe(include='all'))
```

```python
subscribers['signup_date'] = pd.to_datetime(subscribers['signup_date'], format='%Y-%m-%d', errors='coerce')
subscribers['churn_date'] = pd.to_datetime(subscribers['churn_date'], format='%Y-%m-%d', errors='coerce')

# titles 
titles['date_added'] =  pd.to_datetime(titles['date_added'], format='%Y-%m-%d', errors='coerce')
titles['license_expiry'] = pd.to_datetime(titles['license_expiry'], format='%Y-%m-%d', errors='coerce')

# wathch history

watch_history['watch_date'] = pd.to_datetime(watch_history['watch_date'], format='%Y-%m-%d', errors='coerce')

# rating
rating['rating_date'] = pd.to_datetime(rating['rating_date'], format='%Y-%m-%d', errors='coerce')

# reviews
reviews['review_date'] = pd.to_datetime(reviews['review_date'], format='%Y-%m-%d', errors='coerce')

# watchlist
watchlist['added_date'] = pd.to_datetime(watchlist['added_date'], format='%Y-%m-%d', errors='coerce')
```
