```python
import pandas as pd
import numpy as np
```

```python
# load the cleaned datasets

subscribers = pd.read_csv('subscribers_cleaned.csv')
titles = pd.read_csv('titles_cleaned.csv')
watchlist = pd.read_csv('watchlist_cleaned.csv')
watch_history = pd.read_csv('watch_history_cleaned.csv')
rating = pd.read_csv('ratings_cleaned.csv')
reviews = pd.read_csv('reviews_cleaned.csv')
```

```python
# KPI 1
total_watch_hours = watch_history['watch_duration_min'].sum()/60
print(f"Total Watch Hours: {total_watch_hours:,.2f}")
```

```python
# kpi 2
total_subscribers = len(subscribers)
active_subscribers  = subscribers['is_active'].sum()
active_rate =  active_subscribers/total_subscribers * 100
print(f"Total active rate: {active_rate:,.2f}")
```

```python
# kpi 3
inactive_subscribers = total_subscribers - active_subscribers
churn_rate = inactive_subscribers/total_subscribers * 100
print(f"Total churn rate: {churn_rate:,.2f}")
```

```python
# kpi 4
avg_completion_pct = watch_history['completion_pct'].mean()
print(f"Average Completion Percentage: {avg_completion_pct:,.2f}")
```

```python
# kpi 5
active_users = subscribers[subscribers["is_active"] == True]
mrr = active_users["monthly_price_usd"].sum()

print(f"Monthly Recurring Revenue: ${mrr:,.2f}")
```

```python
# kpi 6

arpu = mrr/active_subscribers
print(f"Average Revenue Per User: ${arpu:,.2f}")
```

```python
# kpi 7
avg_watch_time_per_subscriber = total_watch_hours / active_subscribers

print(
    f"Average Watch Time / Active Subscriber: "
    f"{avg_watch_time_per_subscriber:.2f} hours"
)
```

```python
# kpi 8

total_watchlist_entries = len(watchlist)
watched_entries = watchlist["watched"].sum()
watchlist_conversion = (
    watched_entries / total_watchlist_entries
) * 100

print(f"Watchlist Conversion: {watchlist_conversion:.2f}%")
```

```python
# kpi 9

titles_sorted = titles.sort_values(
    "total_plays",
    ascending=False
)

total_titles = len(titles_sorted)
top_10_count = int(np.ceil(total_titles * 0.10))
top_10_titles = titles_sorted.head(top_10_count)
top_10_plays = top_10_titles["total_plays"].sum()

total_plays = titles["total_plays"].sum()

hit_concentration = (
    top_10_plays / total_plays
) * 100

print(f"Hit Concentration: {hit_concentration:.2f}%")
```

```python
# kpi 10

original_titles = titles[titles["is_original"] == True]

original_hours = original_titles["total_watch_hours"].sum()

all_title_hours = titles["total_watch_hours"].sum()

originals_share = (
    original_hours / all_title_hours
) * 100

print(f"Originals Share of Hours: {originals_share:.2f}%")
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Watch Hours",
        "Active Rate",
        "Churn Rate",
        "Avg Completion Rate",
        "Monthly Recurring Revenue",
        "ARPU",
        "Avg Watch Time / Subscriber",
        "Watchlist Conversion",
        "Hit Concentration",
        "Originals Share of Hours"
    ],
    
    "Value": [
        total_watch_hours,
        active_rate,
        churn_rate,
        avg_completion_pct,
        mrr,
        arpu,
        avg_watch_time_per_subscriber,
        watchlist_conversion,
        hit_concentration,
        originals_share
    ]
})
pd.set_option("display.float_format", "{:,.2f}".format)
kpi_summary
```
